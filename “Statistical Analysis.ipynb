```python
# 📊 Statistics Project - Google Colab Version

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Create Sample Dataset
data = pd.DataFrame({
    'StudentID': range(1, 21),
    'Age': np.random.randint(18, 25, 20),
    'Marks': np.random.randint(60, 100, 20),
    'Department': np.random.choice(['CSE', 'EEE', 'MECH'], 20)
})

print("📌 Dataset:\n")
print(data)

# Step 2: Descriptive Statistics
print("\n📊 Descriptive Statistics:")
print("Mean Marks:", np.mean(data['Marks']))
print("Median Marks:", np.median(data['Marks']))
print("Standard Deviation:", np.std(data['Marks']))

# Step 3: Random Sampling
random_sample = data.sample(n=5)
print("\n🎯 Random Sample:\n", random_sample)

# Step 4: Systematic Sampling (every 2nd row)
systematic_sample = data.iloc[::2]
print("\n📌 Systematic Sample:\n", systematic_sample)

# Step 5: Stratified Sampling
stratified_sample = data.groupby('Department').apply(lambda x: x.sample(1))
print("\n📊 Stratified Sample:\n", stratified_sample)

# Step 6: Cluster Sampling (choose CSE department)
cluster_sample = data[data['Department'] == 'CSE']
print("\n📌 Cluster Sample:\n", cluster_sample)

# Step 7: Visualization
plt.hist(data['Marks'])
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```
